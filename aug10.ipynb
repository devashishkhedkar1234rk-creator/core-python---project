#### 1. Relational Operators
#### Boolean value
#### True & False




```python
a=20
b=70

print(a>b)
print(a<b)
print(a==b)
print(a!=b)
print(a>=b)
print(a<=b)
```

#### 2. if conition

```python
age=25

if age>=18:
    print("You are eligible to vote")
```

#### if-else


```python
age=12

if age>=18:

    print("Eligible to vote ")

else:

    print("Not eligible to Vote")
```

#### 4. if-elif-else

```python
marks=95

if marks>=90:
    print("Grade A+")
elif marks>=75:
    print("Grade A")
elif marks>=60:
    print("Grade B")
else:
    print("Grade C")        
```

#### 5. and condition

```python
age=30
salary=30000

if age>=18 and salary >=25000:
    print("Eligible")
else:
    print("Not Eligible")    
```

#### 6. or condition

```python
marks=60
attendance=25

if marks >=75 or attendance >=75:

    print("Eligible")

else:
    print("Not Eligible")
```

#### Combined Example

```python
age=15

marks=85

if age>=18 and marks>=75:
    print("Eligible for admission")
elif age>=18 or marks>=75:
    print("Partially Eligible")
else:
    print("Not Eligible")    
```
