##### Python string File Extension

```python
file="myfile12345.pdf"

count=file.index(".")
name=file[0:count]
print(name.upper())
print(name.title())
print(name.isalnum())

extension=file[count+1:]
print(extension.upper())
print(extension.isalnum())
print(extension.isdigit())
```

#### String methods / String Functions

```python
#title(),capitalize()
name="devashish khedkar"
print(name.title())
print(name.capitalize())
```

```python
#isalnum()
a="Python456"
b="Python 456"
print(a.isalnum())
print(b.isalnum())
```

```python
#isdigit()
a="12345"
b="123abv"
print(a.isdigit())
print(b.isdigit())
```

```python
#isalpha()
a="Python"
b="Python789"
print(a.isalpha())
print(b.isalpha())
```

```python
#File Name & Extension
file="student.pdf"
count=file.index(".")
name=file[:count]
extension=file[count+1:]
print("File Name",name)
print("Extension",extension)
```

```python
#Startswith() & endswith()
file="python.py"
print(file.startswith("python"))
print(file.endswith(".py"))
```

```python
#replace()
name="I Like Java"
print(name.replace("Java","Python"))
```

```python
#position()
text="I am learning Python"
position=text.find("Python")
print(position)
```

```python
#count()
text="banana"
print(text.count("a"))
```

```python
#swapcase
name="PyThOn"
print(name.swapcase())
```
