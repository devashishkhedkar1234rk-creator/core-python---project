### To perform mathematical operations on data we use arithematic operators
##  + , - , * , / , // , % , **

### Addition +

```python
a=10.5
b=20.5
c=a+b
print(a,type(a))
print(b,type(b))
print(c,type(c))
print("Total", c)
```

### Subtraction -

```python
a=50.5
b=20.5
res=a-b
print(a,type(a))
print(b,type(b))
print(res,type(res))
print("Total",res)
```

### Multiplication *

```python
a=12.5
b=4.5
res=a*b
print(a,type(a))
print(b,type(b))
print(res,type(res))
print("Total",res)
```

### Division /

```python
a=45.0
b=5.0
res=a/b
print(a,type(a))
print(b,type(b))
print(res,type(res))
print("Total",res)
```

### Modulus %

```python
a=29
b=6
res=a%b
print(a,type(a))
print(b,type(b))
print(res,type(res))
print("Total",res)
```

### Exponent **

```python
a=3
b=4
res=a**b
print(a,type(a))
print(b,type(b))
print(res,type(res))
print("Total",res)
```

### Floor Division //

```python
a=17
b=5
res=a//b
print(a,type(a))
print(b,type(b))
print(res,type(res))
print("Total",res)
```
